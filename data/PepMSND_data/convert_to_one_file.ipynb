```python
import os
import pandas as pd

files = [f for f in os.listdir(".") if "X_test" in f]
clusters = [f.split("test")[1].split(".")[0] for f in files]

df_list = []

for csv_file in files:
    df = pd.read_csv(csv_file)
    df_list.append(df)

if df_list:
    # Use 'keys' to automatically map the clusters to the rows, then reset the index
    master_df = pd.concat(df_list, keys=clusters, names=["cluster"]).reset_index(level="cluster")
    master_df = master_df.reset_index(drop=True)
else:
    master_df = pd.DataFrame()

print(len(master_df))
```

```python
master_df.to_csv("PepMSND_clustered_data.csv", index=False)
```
